```python
  # Simple Calculator
# Input two numbers and an operator (+, -, *, /).
# Perform the operation using if-else.
# Handle division by zero.
try:


    num1,num2=int(input("enter first number :")),int(input("enter second number :"))
    oprt=input("enter operator(+,-,*,/)")
    if oprt == "+" :
        print(num1+num2)
    elif oprt == "-" :
        print(num1-num2)
    elif oprt == "*":
        print(num1*num2)
    elif oprt == "/":
        print(num1/num2)

except ZeroDivisionError:
      print("STOP ! YOU CAN NOT DIVIDE BY ZERO")
except ValueError:
    print("Please enter a valid number")

```

```python
# Find the sum of first n natural numbers.

num=int(input("enter the number :"))

i=1
while i<2:
    print(num*(num+1)/2)
    i=i+1
```

```python
# Find the factorial of a number.


num=int(input("enter your number :"))

i=1 
fact=i
while i<=num:
    fact=fact * i
    i=i+1    
print(fact)
```

```python
#  Count the number of digits in a number.

num=int(input("enter the number :"))
 
count=0
while num>0:
    count=count+1
    num=num//10

print(count)
```

```python
#  Reverse a number.
num=int(input("enter the number :"))
rev=0
while num > 0 :
    rev=rev * 10 + num% 10 
    num=num//10
print(rev)
```

```python
# Check if a number is a palindrome.

num= int(input("enter the number :"))
fix_n=num
rev=0
while num >0:
    rev=rev*10 + num%10
    num=num//10

if fix_n==rev:
    print("This is palindrome :", rev)
else :
    print("it is not")
```
