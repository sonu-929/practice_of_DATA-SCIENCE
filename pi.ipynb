```python
  # Simple Calculator
# Input two numbers and an operator (+, -, *, /).
# Perform the operation using if-else.
# Handle division by zero.
try:


    num1,num2=int(input("enter first number :")),int(input("enter second number :"))
    oprt=input("enter operator(+,-,*,/)")
    if oprt == "+" :
        print(num1+num2)
    elif oprt == "-" :
        print(num1-num2)
    elif oprt == "*":
        print(num1*num2)
    elif oprt == "/":
        print(num1/num2)

except ZeroDivisionError:
      print("STOP ! YOU CAN NOT DIVIDE BY ZERO")
except ValueError:
    print("Please enter a valid number")

```

```python
# Triangle Checker
# Input three side lengths.
# Check whether they can form a triangle.


# Let the 3 side named as a, b ,c
    
a=int(input("enter the length of a : "))
b=int(input("enter the length of b : "))
c=int(input("enter the length of c : "))

if (a+b > c) and (a+c > b) and (b+c >a):
    print("This is a triangle")
else :
    print("This is not a triangle")
```

```python
# Income Tax Calculator
# Use these tax slabs:
# Income ≤ 250000 → No tax
# 250001–500000 → 5%
# 500001–1000000 → 20%
# Above 1000000 → 30%

income=int(input("Enter your income : "))
if income <= 250000:
    print("You are supposed to not give any TAX")
elif income <= 500000 :
    print(" You have to pay 5% TAX")
elif income <= 1000000 :
    print("You have to pay  20% TAX ")
else :
     print("You have to pay  30% TAX ")


```
